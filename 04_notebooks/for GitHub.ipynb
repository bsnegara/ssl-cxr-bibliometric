```python
%%R
library(readxl)
out <- "repo_export"; dir.create(out, showWarnings = FALSE)

# 1. Keputusan screening (207 rekaman)
s <- read_excel("03_clean/screening.xlsx")
dec <- s[, c("EID", "DOI", "Year", "keputusan", "alasan", "tugas")]
names(dec) <- c("EID", "DOI", "Year", "decision", "code", "task")
write.csv(dec, file.path(out, "screening_decisions.csv"), row.names = FALSE)
print(table(dec$decision))          # cek: I = 119, E = 88

# 2. Sampel validasi (sesuaikan nama file dan kolom tiap putaran)
f <- list.files("03_clean", pattern = "^sampel_r", full.names = TRUE); print(f)
for (i in seq_along(f)) {
  v <- read_excel(f[i])
  keep <- intersect(names(v), c("EID", "keputusan", "keputusan_r2", "alasan_r2"))
  write.csv(v[, keep], file.path(out, sprintf("validation_round%d.csv", i)), row.names = FALSE)
}

# 3. Matriks ekstraksi (hanya isian Bapak)
m <- read_excel("03_clean/matriks_ekstraksi.xlsx")
m <- m[, setdiff(names(m), c("AU", "TI", "AB", "SO"))]
write.csv(m, file.path(out, "extraction_matrix.csv"), row.names = FALSE)
```
